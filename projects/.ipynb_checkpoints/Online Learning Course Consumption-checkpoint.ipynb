# Online Learning Course Consumption Analysis

This notebook analyzes learner behavior using the Kaggle *Online Learning Course Consumption Dataset*.

## Import Libraries

```

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

plt.show()
```

## Load Dataset

```

df = pd.read_csv("online_learning_course_consumption.csv")
df.head()
```

## Dataset Overview

```

df.shape
df.info()
df.describe()
```

## Data Cleaning

```

df.isnull().sum()
df.drop_duplicates(inplace=True)
df.fillna(method='ffill', inplace=True)
```

## Top 10 Most Enrolled Courses

```

df['course_name'].value_counts().head(10).plot(kind='bar')
plt.title("Top 10 Most Enrolled Courses")
plt.xlabel("Course Name")
plt.ylabel("Number of Learners")
plt.show()
```

## Course Category Distribution

```

sns.countplot(data=df, x='category')
plt.title("Course Category Distribution")
plt.xticks(rotation=45)
plt.show()
```

## Course Completion Rate

```

completion_rate = df['course_completed'].value_counts(normalize=True) * 100
completion_rate.plot(kind='pie', autopct='%1.1f%%')
plt.title("Course Completion Rate")
plt.ylabel("")
plt.show()
```

## Time Spent vs Completion

```

sns.boxplot(data=df, x='course_completed', y='time_spent_hours')
plt.title("Time Spent vs Course Completion")
plt.show()
```

## Key Insights
- Higher engagement increases completion
- Some categories perform better than others
- Time spent is a strong indicator of success
